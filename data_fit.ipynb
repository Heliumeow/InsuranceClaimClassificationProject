```python
def dummies_column(column_name,column_list):
    for code in column_list:
        if code not in df.columns: 
            df.insert(df.shape[1],code,0)
        for i in df.index:
            codes=df[column_name][i]
            codes=str(codes)
            if code[len(column_name)+1:] in codes:
                df.loc[i,code]=1

def drop_columns(column_name):
    df.drop(columns=[column_name],inplace=True)
```
