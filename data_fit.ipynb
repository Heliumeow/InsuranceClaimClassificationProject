```python
def dummies_column(column_name,column_list):
    for code in column_list:
        if code not in df.columns: 
            df.insert(df.shape[1],code,0)
        for i in df.index:
            codes=df[column_name][i]
            codes=str(codes)
            if code[len(column_name)+1:] in codes:
                df.loc[i,code]=1

def drop_columns(column_name):
    df.drop(columns=[column_name],inplace=True)
```

```python
dummies_column('BEN_TYPE',BEN_TYPE_list)
drop_columns('BEN_TYPE')
```

```python
dummies_column('MBR_TYPE',MBR_TYPE_list)
drop_columns('MBR_TYPE')
dummies_column('BEN_HEAD',BEN_HEAD_list)
drop_columns('BEN_HEAD')
dummies_column('PROV_LEVEL',PROV_LEVEL_list)
drop_columns('PROV_LEVEL')
```

```python
def value_replace(column_name,divide_list):
    for i in df.index:
        value=df[column_name][i]+1
        new_value=bisect.bisect_left(divide_list,value)
        df.loc[i,column_name]=new_value

```
